```python
from sentence_transformers import SentenceTransformer, util
# Modelo pequeño, se descarga la primera vez (30 seg aprox)
modelo = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
print("Modelo cargado.")
```

```python
# Reemplaza estas palabras con las de tu propio proyecto
# Mezcla conceptos cercanos y lejanos para ver la geometria
palabras = [
    # cluster instituciones
    "superintendencia", "CMF", "AFP", "legal",
    # cluster creadores
    "creadores", "copywriters", "microcopys", "diseño",
    # cluster aprobadores
    "gobernanza", "riesgo/legal", "aprobadores", "normativo",
]

# Comparar cada palabra con las demas
print("MATRIZ DE SIMILITUD (cuanto mas alto, mas parecidos):\n")
print(f"{'':15s}", end="")
for p in palabras[:6]:
    print(f"{p[:8]:>10s}", end="")
print()
for p1 in palabras[:6]:
    print(f"{p1[:14]:15s}", end="")
    for p2 in palabras[:6]:
        emb1 = modelo.encode(p1, convert_to_tensor=True)
        emb2 = modelo.encode(p2, convert_to_tensor=True)
        sim = util.cos_sim(emb1, emb2).item()
        print(f"{sim:>10.2f}", end="")
    print()
```

```python
from transformers import pipeline
clasificador = pipeline("sentiment-analysis",
                       model="nlptown/bert-base-multilingual-uncased-sentiment")
print("Clasificador listo.")
```

```python
frases_a_clasificar = [
    "Este diseno me parece extraordinario, sobrio y memorable.",
    "La interfaz es completamente confusa, no sirve para nada.",
    "El proyecto cumple con lo esperado pero no destaca.",
    # Agrega frases del contexto de tu proyecto:
    "Revisar cumplimiento de lineamientos regulatorios del sector financiero",
    "Copywriting financiero denso",
]

for frase in frases_a_clasificar:
    r = clasificador(frase)[0]
    print(f"  {r['label']} ({r['score']:.0%})  ->  '{frase[:60]}'")
```

```python
generador = pipeline("text-generation",
                     model="datificate/gpt2-small-spanish",
                     max_new_tokens=60)
print("Generador listo.")
```

```python
prompt = "Evalúa la conformidad normativa visual de interfaces financieras: visibilidad de disclaimers, jerarquía tipográfica, casillas de verificación"

# Ejecuta esta celda VARIAS veces y ve como cambia la respuesta
resultado = generador(prompt, num_return_sequences=1,
                     do_sample=True, temperature=0.8)
print(resultado[0]['generated_text'])
```
